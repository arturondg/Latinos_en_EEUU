```python
#COUNTIES
import fiona
import geopandas as gpd

# Configurar el driver explícitamente
fiona.drvsupport.supported_drivers['LIBKML'] = 'rw'

# Leer el archivo KML con GeoPandas
gdf_counties = gpd.read_file('../Files/Cartographic Boundary/Counties/KML/cb_2020_us_county_5m.kml', driver='LIBKML')
print(gdf_counties.head())
```

```python
#STATES
# Configurar el driver explícitamente
fiona.drvsupport.supported_drivers['LIBKML'] = 'rw'

# Leer el archivo KML con GeoPandas
gdf = gpd.read_file('../Files/Cartographic Boundary/States/KML/cb_2020_us_state_5m.kml', driver='LIBKML')
print(gdf.head())
```

```python
dtypes = {
    'FILEID': str, 'SUMLEV': str, 'STUSAB': str, 'LOGRECNO': str, 'STATENS': str,
    'COUNTYNS': str, 'COUSUBNS': str, 'SUBMCDNS': str, 'CONCITNS': str, 
    'PLACENS': str, 'ZCTA': str, 'BASENAME': str, 'NAME': str, 'POP100': int,
    'HU100': int, 'INTPTLAT': float, 'INTPTLON': float, 'LSADC': str, 'P0020001': int,
    'P0020002': int
}

dtypes_counties = {
    'FILEID': str, 'SUMLEV': str, 'STUSAB': str, 'LOGRECNO': str, 'STATENS': str,
    'COUNTYNS': str, 'COUSUBNS': str, 'SUBMCDNS': str, 'CONCITNS': str, 
    'PLACENS': str, 'ZCTA': str, 'BASENAME': str, 'NAME': str, 'POP100': int,
    'HU100': int, 'INTPTLAT': float, 'INTPTLON': float, 'LSADC': str, 'P0020001': int,
    'P0020002': int, 'STATENAME': str
}
```

```python
#COUNTIES
import pandas as pd

# Leer la base de datos de los estados
df_counties = pd.read_excel('../Files/DataBases [COMPLETE]/3.1.-EEUU_States-Counties.xlsx', dtype=dtypes_counties)

# Verificar las primeras filas
print(df_counties.head())
```

```python
#STATES
# Leer la base de datos de los estados
df_states = pd.read_excel('../Files/DataBases [COMPLETE]/2.-EEUU_States.xlsx', dtype=dtypes)

# Verificar las primeras filas
print(df_states.head())
```

```python
#COUNTIES
# Unir los datos
merged_gdf_counties = gdf_counties.merge(df_counties, left_on='COUNTYNS', right_on='COUNTYNS')

# Verificar las primeras filas del GeoDataFrame combinado
print(merged_gdf_counties.head())
```

```python
#SATES
# Unir los datos
merged_gdf_states = gdf.merge(df_states, left_on='STATENS', right_on='STATENS')

# Verificar las primeras filas del GeoDataFrame combinado
print(merged_gdf_states.head())
```

```python
import folium
from branca.colormap import LinearColormap, StepColormap

# Crear el mapa base
m = folium.Map(location=[37.8, -96], zoom_start=4)

# Crear la primera capa con el colormap lineal
# -------------------------------------------------
vmin_value = merged_gdf_states['P0020002'].min()
vmax_value = merged_gdf_states['P0020002'].max()

# Definir una gama de colores en tonos de azul
blue_colors = [
    '#f7fbff',  # azul muy claro
    '#deebf7',  # azul claro
    '#c6dbef',  # azul medio-claro
    '#9ecae1',  # azul más fuerte
    '#6baed6',  # azul fuerte
    '#4292c6',  # azul muy fuerte
    '#2171b5',  # azul oscuro
    '#08519c',  # azul muy oscuro
    '#08306b'   # casi negro
]

# Crear el colormap lineal con colores y rango basado en tus datos
linear = LinearColormap(blue_colors, vmin=vmin_value, vmax=vmax_value)
linear = linear.to_step(
    n=9,
    data=[15504, 273493, 531483, 789472, 1047462, 1305451, 1563441, 1821430, 2079420, 2337410],
    method='quantiles',
    round_method='int'
)

# Añadir una nueva columna con los colores basados en el colormap
merged_gdf_states['color'] = merged_gdf_states['P0020002'].apply(lambda x: linear(x))

# Crear un grupo de capas para el primer mapa
layer1 = folium.FeatureGroup(name='Hispanos/Latinos por Estado (Azul)')

# Añadir GeoJson al grupo de capas con colores personalizados
folium.GeoJson(
    merged_gdf_states,
    style_function=lambda feature: {
        'fillColor': feature['properties']['color'],
        'color': 'black',
        'weight': 2,
        'dashArray': '5, 5',
        'fillOpacity': 0.8
    },
    tooltip=folium.features.GeoJsonTooltip(
        fields=['NAME', 'P0020002'],
        aliases=['Estado', 'Cantidad de Hispanos/Latinos'],
        localize=True
    )
).add_to(layer1)

# Crear la segunda capa con el StepColormap
# -----------------------------------------
# Define los colores y los intervalos de valores
colors = [
    '#fff5f0',  # rojo muy claro
    '#fee0d2',  # rojo claro
    '#fc9272',  # rojo más fuerte
    '#fb6a4a',  # rojo fuerte
    '#ef3b2c',  # rojo muy fuerte
    '#a50f15',  # rojo muy oscuro
    '#67000d',  # casi negro
    '#190003',  # negro muy oscuro
    '#050001'
]

# Define los intervalos correspondientes para los colores
data_intervals = [1, 100, 500, 1000, 5000, 10000, 50000, 100000, 1000000]

# Crear el StepColormap con colores y rangos definidos
step_colormap = StepColormap(colors, vmin=min(data_intervals), vmax=max(data_intervals), index=data_intervals)

# Añadir una nueva columna con los colores basados en el colormap
merged_gdf_counties['color'] = merged_gdf_counties['P0020002'].apply(lambda x: step_colormap(x))

# Crear un grupo de capas para el segundo mapa
layer2 = folium.FeatureGroup(name='Hispanos/Latinos por Condado (Rojo)')

# Añadir GeoJson al grupo de capas con colores personalizados
folium.GeoJson(
    merged_gdf_counties,
    style_function=lambda feature: {
        'fillColor': feature['properties']['color'],
        'color': 'black',
        'weight': 2,
        'dashArray': '5, 5',
        'fillOpacity': 0.9
    },
    tooltip=folium.features.GeoJsonTooltip(
        fields=['STATENAME', 'NAME', 'P0020002'],
        aliases=['Estado', 'County', 'Cantidad de Hispanos/Latinos'],
        localize=True
    )
).add_to(layer2)

# Añadir un título en un div fijo
title_html = '''
             <div style="position: fixed; 
                         top: 10px; left: 50px; width: 100%;
                         text-align: center; font-size: 20px; z-index:9999;">
                         <b>Hispanos/Latinos en Estados Unidos</b>
             </div>
             '''
m.get_root().html.add_child(folium.Element(title_html))

# Añadir créditos de autor
credit_html = '''
             <div style="position: fixed; 
                         bottom: 10px; left: 50px; width: 100%;
                         text-align: left; font-size: 14px; z-index:9999;">
                         <b>Elaborado por: Arturo Nicolás Díaz García</b>
             </div>
             '''
m.get_root().html.add_child(folium.Element(credit_html))

# Añadir las capas al mapa
layer1.add_to(m)
layer2.add_to(m)

# Añadir control de capas
folium.LayerControl().add_to(m)

# Añadir leyenda para el primer mapa
linear.caption = 'Cantidad de Hispanos/Latinos por Estado'
linear.add_to(m)

# Crear y añadir leyenda para el segundo mapa
legend_html = '''
             <div style="position: fixed; 
                         bottom: 50px; left: 10px; width: 250px; height: auto; 
                         background-color: white; padding: 10px; border: 2px solid black; 
                         z-index:9999;">
                         <h4 style="margin-top: 0;">Número de Hispanos/Latinos por Counties:</h4>
                         <div style="display: flex; flex-direction: column;">
'''

for i in range(len(colors)):
    legend_html += f'''
        <div style="display: flex; align-items: center; margin-bottom: 5px;">
            <div style="background-color: {colors[i]}; width: 30px; height: 15px; border: 1px solid black; margin-right: 5px;"></div>
            <span>{data_intervals[i]} - {data_intervals[i + 1] if i + 1 < len(data_intervals) else 'Más de ' + str(data_intervals[i])}</span>
        </div>
    '''

legend_html += '''
                         </div>
             </div>
             '''

# Añadir la leyenda al mapa
m.get_root().html.add_child(folium.Element(legend_html))

# Guardar el mapa
m.save('Latinos_en_EEUU_por_Condado[COMPLETO_CON_LAYERS].html')
```
