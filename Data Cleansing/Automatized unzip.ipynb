```python
import os
import zipfile

# Ruta del directorio que contiene los archivos ZIP
zip_directory = 'Archives'

# Rutas de los directorios de destino
dest_directories = {
    'geoheader': 'DataBases/2020 PL Geoheader',
    'segment1': 'DataBases/2020 PL Segment 1',
    'segment2': 'DataBases/2020 PL Segment 2',
    'segment3': 'DataBases/2020 PL Segment 3'
}

# Crear directorios de destino si no existen
for path in dest_directories.values():
    os.makedirs(path, exist_ok=True)

# Función para determinar el directorio de destino
def get_destination_directory(filename):
    if 'geo' in filename:
        return dest_directories['geoheader']
    elif '00001' in filename:
        return dest_directories['segment1']
    elif '00002' in filename:
        return dest_directories['segment2']
    elif '00003' in filename:
        return dest_directories['segment3']
    return None

# Procesar cada archivo ZIP en el directorio
for zip_filename in os.listdir(zip_directory):
    if zip_filename.endswith('.zip'):
        with zipfile.ZipFile(os.path.join(zip_directory, zip_filename), 'r') as zip_ref:
            zip_ref.extractall(zip_directory)
            
        # Mover y renombrar los archivos descomprimidos
        for filename in zip_ref.namelist():
            destination_directory = get_destination_directory(filename)
            if destination_directory:
                src_path = os.path.join(zip_directory, filename)
                dst_path = os.path.join(destination_directory, os.path.splitext(filename)[0] + '.txt')
                os.rename(src_path, dst_path)
```
