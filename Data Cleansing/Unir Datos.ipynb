```python
import pandas as pd
import os

# Directorios de las carpetas
carpeta1 = 'DataBases/2020 PL Geoheader'
carpeta2 = 'DataBases/2020 PL Segment 1'
output_folder = 'DataBases [WRANGLING]'

# Lista de archivos en ambas carpetas
archivos_carpeta1 = [f for f in os.listdir(carpeta1) if f.endswith('.txt')]
archivos_carpeta2 = [f for f in os.listdir(carpeta2) if f.endswith('.txt')]

# Define los tipos de datos
dtypes_carpeta1 = {
    'FILEID': str, 'SUMLEV': str, 'STUSAB': str, 'LOGRECNO': str, 'STATENS': str,
    'COUNTYNS': str, 'COUSUBNS': str, 'SUBMCDNS': str, 'CONCITNS': str, 
    'PLACENS': str, 'ZCTA': str, 'BASENAME': str, 'NAME': str, 'POP100': int,
    'HU100': int, 'INTPTLAT': float, 'INTPTLON': float, 'LSADC': str
}

dtypes_carpeta2 = {
    'LOGRECNO': str, 'P0020001': int, 'P0020002': int
}
```

```python
# Columnas a filtrar en cada carpeta
columnas_carpeta1 = ['FILEID', 'SUMLEV', 'STUSAB', 'LOGRECNO', 'STATENS',
                     'COUNTYNS', 'COUSUBNS', 'SUBMCDNS', 'CONCITNS', 'PLACENS',
                     'ZCTA', 'BASENAME', 'NAME', 'POP100', 'HU100', 'INTPTLAT',
                     'INTPTLON', 'LSADC']
columnas_carpeta2 = ['LOGRECNO', 'P0020001', 'P0020002']

# Valores a filtrar en la columna SUMLEV
valores_sumlev = ['160', '155', '050', '060', '040']

for archivo1 in archivos_carpeta1:
    # Extraer los primeros dos dígitos del nombre del archivo para buscar en la otra carpeta
    prefix = archivo1[:2]

    # Encontrar el archivo correspondiente en la segunda carpeta
    archivo2 = [f for f in archivos_carpeta2 if f.startswith(prefix)]
    if not archivo2:
        continue
    archivo2 = archivo2[0]

    # Cargar los archivos con el delimitador y encoding especificados
    df1 = pd.read_csv(os.path.join(carpeta1, archivo1), delimiter='|', dtype=dtypes_carpeta1, usecols=columnas_carpeta1, encoding='latin-1')
    df2 = pd.read_csv(os.path.join(carpeta2, archivo2), delimiter='|', dtype=dtypes_carpeta2, usecols=columnas_carpeta2, encoding='latin-1')

    # Filtrar los valores deseados en la columna SUMLEV
    df1_filtrado = df1[df1['SUMLEV'].isin(valores_sumlev)]

    # Unir los DataFrames por 'LOGRECNO'
    df_combinado = pd.merge(df1_filtrado, df2, on='LOGRECNO', how='inner')

    # Guardar el resultado en el destino especificado
    output_file = os.path.join(output_folder, f'combinado_{prefix}.csv')
    df_combinado.to_csv(output_file, index=False, encoding='latin-1')

    # Imprimir mensaje de éxito
    print(f'Archivo combinado guardado exitosamente: {output_file}')
```
